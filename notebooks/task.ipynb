```python
import psycopg2
import pandas as pd

# Connect to your PostgreSQL database
conn = psycopg2.connect(
dbname="bank_reviews",
user="postgres",
password="senu65946655",
host="localhost",
port=5432
)
cur = conn.cursor()
print("Connected to PostgreSQL!")


```

```python
csv_path = r"C:\Users\jkk\Downloads\Customer-Experience\data\task2_output.csv"
df = pd.read_csv(csv_path)
print("CSV Loaded!")
print(df.head())
```

```python
cur.execute("SELECT bank_id, bank_name FROM banks;")
bank_map = {name: bid for bid, name in cur.fetchall()}
print("Banks loaded from DB:", bank_map)
```

```python
row_count = 0

for _, row in df.iterrows():
    bank_name = row["bank"]
    
    if bank_name not in bank_map:
        print(f"Skipping — bank not found in DB: {bank_name}")
        continue
    
    bank_id = bank_map[bank_name]
    
    cur.execute(
        """
        INSERT INTO reviews (
            bank_id, review_text, rating, review_date,
            sentiment_label, sentiment_score, source
        ) VALUES (%s, %s, %s, %s, %s, %s, %s);
        """,
        (
            bank_id,
            row["review_text"],
            row["rating"],
            row["review_date"],
            row["sentiment_label"],
            row["sentiment_score"],
            row["source"]
        )
    )
    
    row_count += 1

conn.commit()
print(f"{row_count} reviews inserted successfully!")
```

```python
cur.close()
conn.close()
print("Database connection closed!")
```
